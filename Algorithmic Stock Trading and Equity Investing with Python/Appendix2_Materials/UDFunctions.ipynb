### Positional Arguments vs. Keyword Arguments

````python
def future_value(pv, rate, nper):
    fv = pv * (1 + rate)**nper
    return fv
````

````python
future_value(100, 0.04, 5) #positional arguments only 
````

````python
future_value(0.04, 100, 5) #positional arguments only - sequence/positions matter!
````

##### we see that the order of the arguments matters when using positional arguments.<br> 
##### The first argument is treated as `pv`, the second as `rate`, and the third as `nper`.

````python
future_value(pv = 100 , rate = 0.04 , nper = 5) # keyword arguments only
````

````python
future_value(rate = 0.04 , nper = 5 , pv = 100) # keyword arguments only - sequence does not matter!
````

````python
future_value(100 , rate = 0.04 , nper = 5) #combination 
````

````python
future_value(100 , 0.04, nper = 5) #combination 
````

### The default argument None

````python
def future_value5(pv, rate, nyears, m = None):
    '''It means if we already have a value for m, then we will use that value. If we do not have a value for m (m = None), then its boolean value will be false, so 
        then we will deliberately start the if statement to commence with not m because not m is true which equivalent to if m is false, then we will set it to 1.
        So: if n = None , then bool(n) = False, then not n = True, then we will set m = 1. 
        If m is already set to a value, then bool(m) = True, then not m = False, then we will not set m to 1.
    '''
    if not m:
        m = 1
        fv = pv * (1 + rate/m)**(nyears*m)
        return fv
````

````python
future_value5(100, 0.04, 5)
````

````python
future_value5(100, 0.04, 5, 12)
````

def npv_irr(rate, values, guess = 0.05)->tuple:
    '''
    Calculate NPV and IRR for a series of cash flows at a given rate of return. 
    
    Parameters
    ----------
    rate : float
        Required Rate of return
    values : list
        List of cash flow values
    guess : float
        Initial guess for IRR : default = 0.05 which says that the IRR(Internal Rate of Return) is 5%.
        
    Returns
    -------
    NPV : float
        Net present value
    IRR : float
        Internal rate of return
    '''   
    # first step: calculate NPV 
    NPV = 0
    for i in range(len(values)):
        NPV += values[i] / (1 + rate)**(i)

    # second step: calculate IRR regarding the tolerance of 0.001 & step of 0.0000001.    
    step = 0.0000001
    target_npv = 0
    tolerance = 0.001 

    while True:
        f = 1 + guess
        npv = 0
        for i in range(len(values)):
            npv += values[i] / f**(i)
        diff = npv - target_npv

        if abs(diff) > tolerance:
            if diff < 0:
                guess -= step
            elif diff > 0:
                guess += step
        else:
            break
    
    return NPV, guess

````python
# my suggestion to mend some part of the above code:
def npv_irr(rate, values, guess = 0.05)->tuple:
    '''
    Calculate NPV and IRR for a series of cash flows at a given rate of return. 
    
    Parameters
    ----------
    rate : float
         Reqired Rate of return : default = 0.06
    values : list
        List of cash flow values
    guess : float
        Initial guess for IRR : default = 0.05 which says that the IRR(Internal Rate of Return) is 5%.
        
    Returns
    -------
    NPV : float
        Net present value
    IRR : float
        Internal rate of return
    '''   
    # first step: calculate NPV 
    NPV = 0
    for i in range(len(values)):
        NPV += values[i] / (1 + rate)**(i)

    # second step: calculate IRR regarding the tolerance of 0.001 & step of 0.0000001. We assume the npv is zero. 
    step = 0.0000001
    tolerance = 0.001 

    while True:
        f = 1 + guess
        npv = 0
        for i in range(len(values)):
            npv += values[i] / f**(i)

        if abs(npv) > tolerance: # it means that the npv is not close enough to zero. Then we need to adjust the guess.
            if npv < 0: # Then we need to increase the guess to get closer to zero.
                guess -= step
            elif npv > 0: # Then we need to decrease the guess to get closer to zero.
                guess += step
        else:
            break
    
    return NPV, "{:.4f}%".format(guess * 100)
````

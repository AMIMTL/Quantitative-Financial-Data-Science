## Finding the optimal SMA Strategy

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("seaborn-v0_8")


%config InlineBackend.figure_format = "retina"
```

```python
def run_strategy(SMA:tuple)->pd.DataFrame:
    data = df.copy()
    data["SMA_S"] = data.price.rolling(window = int(SMA[0])).mean()
    data["SMA_L"] = data.price.rolling(window = int(SMA[1])).mean()
    data["position"] = np.where(data["SMA_S"]>data["SMA_L"] , +1 , -1)
    
    data["log_ret"] = np.log(data.price.div(data.price.shift(1)))  
    data["strategy"] = data.position.shift(1) * data["log_ret"]
    data = data.dropna()

    performance = np.exp(data.iloc[:,-2:].agg({"sum"}))
    
    return performance
```

### Since `brute()` minimize a function and we want to increate the return so we have to put -run_strategy function in the brute function:
`brute(-run_strategy)` ~ minimization of -run_strategy ~ maximization of run_strategy. 
In practice we need to redefince the run_strategy function and multiply its return by -1.

```python
def run_strategy(SMA:tuple)->pd.DataFrame:
    data = df.copy()
    data["SMA_S"] = data.price.rolling(window = int(SMA[0])).mean()
    data["SMA_L"] = data.price.rolling(window = int(SMA[1])).mean()
    data["position"] = np.where(data["SMA_S"]>data["SMA_L"] , +1 , -1)
    
    data["log_ret"] = np.log(data.price.div(data.price.shift(1)))  
    data["strategy"] = data.position.shift(1) * data["log_ret"]
    data = data.dropna()

    performance = np.exp(data.iloc[:,-2:].agg({"sum"}))
    
    return -1 * performance.iloc[0,1]
```

```python
def annual_perf_runned_strategy(SMA:tuple)->pd.DataFrame:
    data = df.copy()
    data["SMA_S"] = data.price.rolling(window = int(SMA[0])).mean()
    data["SMA_L"] = data.price.rolling(window = int(SMA[1])).mean()
    data["position"] = np.where(data["SMA_S"]>data["SMA_L"] , +1 , -1)
    
    data["log_ret"] = np.log(data.price.div(data.price.shift(1)))  
    data["strategy"] = data.position.shift(1) * data["log_ret"]
    data = data.dropna()

    # Annual Performance:
    ann_mean = data.iloc[:,-2:].agg({"mean"}) * 252
    ann_std =  data.iloc[:,-2:].agg({"std"}) * np.sqrt(252)
    
    return pd.concat([ann_mean , ann_std] , axis = 0).T.style.format("{:.2%}")
```

```python
df1 = annual_perf_runned_strategy((46,137))
df1
```

```python
def data_runned_strategy(SMA:tuple)->pd.DataFrame:
    data = df.copy()
    data["SMA_S"] = data.price.rolling(window = int(SMA[0])).mean()
    data["SMA_L"] = data.price.rolling(window = int(SMA[1])).mean()
    data["position"] = np.where(data["SMA_S"]>data["SMA_L"] , +1 , -1)
    
    data["log_ret"] = np.log(data.price.div(data.price.shift(1)))  
    data["strategy"] = data.position.shift(1) * data["log_ret"]

    data["cum_log_ret"] = np.exp(data["log_ret"].cumsum())
    data["cum_strategy"] = np.exp(data["strategy"].cumsum())
    
    data = data.dropna()   
    return data.iloc[:,-2:]
```

## Testing a Strategy

```python
# to Update our code to add instances to the SMA class:
class SMABacktester():
    def __init__(self , symbol:str , SMA_S:int , SMA_L:int , start:str , end:str):
        self.symbol = symbol
        self.SMA_S = SMA_S
        self.SMA_L = SMA_L
        self.start = start
        self.end = end
        self.results = None
        self.get_data()

    def get_data(self):
        raw = pd.read_csv("forex_pairs.csv" , parse_dates=["Date"] , index_col=["Date"])
        raw = raw[self.symbol].to_frame().dropna()
        raw = raw.loc[self.start : self.end].copy()
        raw.columns = ["price"]
        raw["SMA_S"] = raw["price"].rolling(window = self.SMA_S).mean()
        raw["SMA_L"] = raw["price"].rolling(window = self.SMA_L).mean()
        self.data = raw
        return raw

    def set_parameters(self , new_SMA_S = None, new_SMA_L = None):
        if new_SMA_S is not None:
            self.SMA_S = new_SMA_S
            self.data["SMA_S"] = self.data["price"].rolling(window = self.SMA_S).mean()

        if new_SMA_L is not None:
            self.SMA_L = new_SMA_L        
            self.data["SMA_L"] = self.data["price"].rolling(window = self.SMA_L).mean()

    def trading_strategy_performance(self):
        data = self.data.dropna().copy()
        data["position"] = np.where(data["SMA_S"] > data["SMA_L"] , +1 , -1)
        data["log_ret"] = np.log(data["price"].div(data["price"].shift(1)))
        data["strategy"] = data["position"].shift(1).multiply(data["log_ret"])
        data["cum_log_ret"] = np.exp(data["log_ret"].cumsum())
        data["cum_strategy"] = np.exp(data["strategy"].cumsum())

        perf = data["cum_strategy"].iloc[-1] # absolute performance
        cum_log_ret_perf = data["cum_log_ret"].iloc[-1] # Performance of the cummulative log returns
        out_perf = perf - cum_log_ret_perf  # out performance
        
        #pd.options.display.float_format = '{:.2f}%'.format
        KPI = pd.DataFrame({"Absolute Performance of Strategy": [perf] ,
                            "Cummulative Log Returns Performance" : [cum_log_ret_perf] ,
                            "Out Performance": [out_perf]} , 
                            index = ["results {}|{}".format(self.start , self.end)])
        
        KPI = KPI.T.style.format("{:.2%}")
        self.results = data
        
        return KPI
```

```python
# to Update our code to add instances to the SMA class:
class SMABacktester():
    def __init__(self , symbol:str , SMA_S:int , SMA_L:int , start:str , end:str):
        self.symbol = symbol
        self.SMA_S = SMA_S
        self.SMA_L = SMA_L
        self.start = start
        self.end = end
        self.results = None
        self.get_data()

    def get_data(self):
        raw = pd.read_csv("forex_pairs.csv" , parse_dates=["Date"] , index_col=["Date"])
        raw = raw[self.symbol].to_frame().dropna()
        raw = raw.loc[self.start : self.end].copy()
        raw.columns = ["price"]
        raw["SMA_S"] = raw["price"].rolling(window = self.SMA_S).mean()
        raw["SMA_L"] = raw["price"].rolling(window = self.SMA_L).mean()
        self.data = raw
        return raw

    def set_parameters(self , new_SMA_S = None, new_SMA_L = None):
        if new_SMA_S is not None:
            self.SMA_S = new_SMA_S
            self.data["SMA_S"] = self.data["price"].rolling(window = self.SMA_S).mean()

        if new_SMA_L is not None:
            self.SMA_L = new_SMA_L        
            self.data["SMA_L"] = self.data["price"].rolling(window = self.SMA_L).mean()

    def trading_strategy_performance_tabular(self):
        data = self.data.dropna().copy()
        data["position"] = np.where(data["SMA_S"] > data["SMA_L"] , +1 , -1)
        data["log_ret"] = np.log(data["price"].div(data["price"].shift(1)))
        data["strategy"] = data["position"].shift(1).multiply(data["log_ret"])
        data["cum_log_ret"] = np.exp(data["log_ret"].cumsum())
        data["cum_strategy"] = np.exp(data["strategy"].cumsum())

        perf = data["cum_strategy"].iloc[-1] # absolute performance
        cum_log_ret_perf = data["cum_log_ret"].iloc[-1] # Performance of the cummulative log returns
        out_perf = perf - cum_log_ret_perf  # out performance
        
        #pd.options.display.float_format = '{:.2f}%'.format
        self.KPI = pd.DataFrame({"Absolute Performance of Strategy": [perf] ,
                            "Cummulative Log Returns Performance" : [cum_log_ret_perf] ,
                            "Out Performance": [out_perf]} , 
                            index = ["results {}|{}".format(self.SMA_S , self.SMA_L)])
        
        self.KPI = self.KPI.T.style.format("{:.2%}")
        self.results = data        
        return self.KPI

    def trading_strategy_performance_values(self):
        data = self.data.dropna().copy()
        data["position"] = np.where(data["SMA_S"] > data["SMA_L"] , +1 , -1)
        data["log_ret"] = np.log(data["price"].div(data["price"].shift(1)))
        data["strategy"] = data["position"].shift(1).multiply(data["log_ret"])
        data["cum_log_ret"] = np.exp(data["log_ret"].cumsum())
        data["cum_strategy"] = np.exp(data["strategy"].cumsum())

        perf = data["cum_strategy"].iloc[-1] # absolute performance
        cum_log_ret_perf = data["cum_log_ret"].iloc[-1] # Performance of the cummulative log returns
        out_perf = perf - cum_log_ret_perf  # out performance
        
        #pd.options.display.float_format = '{:.2f}%'.format
        self.KPI = pd.DataFrame({"Absolute Performance of Strategy": [perf] ,
                            "Cummulative Log Returns Performance" : [cum_log_ret_perf] ,
                            "Out Performance": [out_perf]} , 
                            index = ["results {}|{}".format(self.SMA_S , self.SMA_L)])
        self.KPI = self.KPI.T
        self.results = data        
        return self.KPI.loc["Absolute Performance of Strategy",:].values

    def plot_results(self):
        if self.results is None:
            print("There is nothing to display! Run a Strategy :-)")
        else:
            self.results.iloc[:,-2:].plot()
            plt.title(f"Trading Strategy VS Buy&Hold Strategy \n Ticker: {self.symbol} | SMA_S: {self.SMA_S} | SMA_L: {self.SMA_L}")
            plt.xlabel("Date")
            plt.ylabel("Value")
            plt.tight_layout()
            plt.autoscale(enable = True, 
                          axis = "x", 
                          tight = True)
            plt.legend(labels = ["Buy&Hold Strategy" , "Trading Strategy"], 
                       bbox_to_anchor = (1,1) ,
                       facecolor = "peachpuff", 
                       frameon = True ,
                       fancybox = True , 
                       shadow = True)
            plt.show()

    def update_and_run_strategy_performance_table(self , SMA:tuple):
        self.set_parameters(int(SMA[0]) , int(SMA[1]))
        return self.trading_strategy_performance_tabular()

    def update_and_run_strategy_performance_value(self , SMA:tuple):
        self.set_parameters(int(SMA[0]) , int(SMA[1]))
        return self.trading_strategy_performance_values()

    def optimization_trading_parameters(self , SMAS_tuple:tuple , SMAL_tuple:tuple):
        ranges = (SMAS_tuple , SMAL_tuple)
        obj_func  = lambda x : -1*self.update_and_run_strategy_performance_value(x)
        optimized_values = brute(obj_func , ranges)
        return optimized_values  ,   self.update_and_run_strategy_performance_value(optimized_values)
```

```python
# to Update our code to add instances to the SMA class:
class SMABacktester():
    ''' Class for the vectorized backtesting of SMA-based trading strategies:

    Attributes
    ==========
    symbol: str
        ticker symbol with which to work with
    SMA_S: int
        time window in days for shorter SMA
    SMA_L: int
        time window in days for longer SMA
    start: str
        start date for data retrieval
    end: str
        end date for data retrieval
        
    Methods
    =======
    get_data:
        retrieves and prepares the data
        
    set_parameters:
        sets one or two new SMA parameters
        
    trading_strategy_performance_tabular:
        runs the backtest for the SMA-based strategy and return the KPI in a dataframe.

    trading_strategy_performance_values:
        runs the backtest for the SMA-based strategy and return the KPIs values.

    update_and_run_strategy_performance_table:
        Update the SMA parameter in the backtest trading strategy and display in a dataframe.

    update_and_run_strategy_performance_value:
        Update the SMA parameter in the backtest trading strategy and display the values.
         
    plot_results:
        plots the performance of the strategy compared to buy and hold.

    optimization_trading_parameters:
        compute the argmin of the trading strategy within a range of SMA's parameters. 
        '''
    def __init__(self , symbol:str , SMA_S:int , SMA_L:int , start:str , end:str):
        self.symbol = symbol
        self.SMA_S = SMA_S
        self.SMA_L = SMA_L
        self.start = start
        self.end = end
        self.results = None
        self.get_data()
        
    def __repr__(self):
        rep = "SMABacktester(symbol = {}, SMA_S = {}, SMA_L = {}, start = {}, end = {})"
        return rep.format(self.symbol, self.SMA_S, self.SMA_L, self.start, self.end)     


    def get_data(self):
        '''Retreives and Prepares the data.
        '''
        raw = pd.read_csv("forex_pairs.csv" , parse_dates=["Date"] , index_col=["Date"])
        raw = raw[self.symbol].to_frame().dropna()
        raw = raw.loc[self.start : self.end].copy()
        raw.columns = ["price"]
        raw["SMA_S"] = raw["price"].rolling(window = self.SMA_S).mean()
        raw["SMA_L"] = raw["price"].rolling(window = self.SMA_L).mean()
        self.data = raw
        return raw

    def set_parameters(self , new_SMA_S = None, new_SMA_L = None):
        '''Updates SMA parameters and resp. time series.
        '''
        if new_SMA_S is not None:
            self.SMA_S = new_SMA_S
            self.data["SMA_S"] = self.data["price"].rolling(window = self.SMA_S).mean()

        if new_SMA_L is not None:
            self.SMA_L = new_SMA_L        
            self.data["SMA_L"] = self.data["price"].rolling(window = self.SMA_L).mean()

    def trading_strategy_performance_tabular(self):
        '''Display the performance of the trading strategy vs Buy&Hold in a dataframe.
        '''
        data = self.data.dropna().copy()
        data["position"] = np.where(data["SMA_S"] > data["SMA_L"] , +1 , -1)
        data["log_ret"] = np.log(data["price"].div(data["price"].shift(1)))
        data["strategy"] = data["position"].shift(1).multiply(data["log_ret"])
        data["cum_log_ret"] = np.exp(data["log_ret"].cumsum())
        data["cum_strategy"] = np.exp(data["strategy"].cumsum())

        perf = data["cum_strategy"].iloc[-1] # absolute performance
        cum_log_ret_perf = data["cum_log_ret"].iloc[-1] # Performance of the cummulative log returns
        out_perf = perf - cum_log_ret_perf  # out performance
        
        #pd.options.display.float_format = '{:.2f}%'.format
        self.KPI = pd.DataFrame({"Absolute Performance of Strategy": [perf] ,
                            "Cummulative Log Returns Performance" : [cum_log_ret_perf] ,
                            "Out Performance": [out_perf]} , 
                            index = ["results {}|{}".format(self.SMA_S , self.SMA_L)])
        
        self.KPI = self.KPI.T.style.format("{:.2%}")
        self.results = data        
        return self.KPI

    def trading_strategy_performance_values(self):
        '''Display the values of the cummulative performance of the trading strategy compared to Buy & Hold strategy.
        '''
        data = self.data.dropna().copy()
        data["position"] = np.where(data["SMA_S"] > data["SMA_L"] , +1 , -1)
        data["log_ret"] = np.log(data["price"].div(data["price"].shift(1)))
        data["strategy"] = data["position"].shift(1).multiply(data["log_ret"])
        data["cum_log_ret"] = np.exp(data["log_ret"].cumsum())
        data["cum_strategy"] = np.exp(data["strategy"].cumsum())

        perf = data["cum_strategy"].iloc[-1] # absolute performance
        cum_log_ret_perf = data["cum_log_ret"].iloc[-1] # Performance of the cummulative log returns
        out_perf = perf - cum_log_ret_perf  # out performance
        
        #pd.options.display.float_format = '{:.2f}%'.format
        self.KPI = pd.DataFrame({"Absolute Performance of Strategy": [perf] ,
                            "Cummulative Log Returns Performance" : [cum_log_ret_perf] ,
                            "Out Performance": [out_perf]} , 
                            index = ["results {}|{}".format(self.SMA_S , self.SMA_L)])
        self.KPI = self.KPI.T
        self.results = data        
        return self.KPI.loc["Absolute Performance of Strategy",:].values

    def plot_results(self):
        '''Plot the cummulative returns of the trading strategy vs the cummulative log returns of the buy&hold strategy.
        '''
        if self.results is None:
            print("There is nothing to display! Run a Strategy :-)")
        else:
            self.results.iloc[:,-2:].plot()
            plt.title(f"Trading Strategy VS Buy&Hold Strategy \n Ticker: {self.symbol} | SMA_S: {self.SMA_S} | SMA_L: {self.SMA_L}")
            plt.xlabel("Date")
            plt.ylabel("Value")
            plt.tight_layout()
            plt.autoscale(enable = True, 
                          axis = "x", 
                          tight = True)
            plt.legend(labels = ["Buy&Hold Strategy" , "Trading Strategy"], 
                       bbox_to_anchor = (1,1) ,
                       facecolor = "peachpuff", 
                       frameon = True ,
                       fancybox = True , 
                       shadow = True)
            plt.show()

    def update_and_run_strategy_performance_table(self , SMA:tuple):
        '''Update the SMA parameter in the backtest trading strategy and return the dataframe of the trading strategy within the new parameters.
        '''
        self.set_parameters(int(SMA[0]) , int(SMA[1]))
        return self.trading_strategy_performance_tabular()

    def update_and_run_strategy_performance_value(self , SMA:tuple):
        '''Update the SMA parameter in the backtest trading strategy and return the value of the trading strategy within the new parameters.
           It will be used to find the final optimal value by using the brute function from scipy.optimize
        '''
        self.set_parameters(int(SMA[0]) , int(SMA[1]))
        return self.trading_strategy_performance_values()

    def optimization_trading_parameters(self , SMAS_tuple:tuple , SMAL_tuple:tuple):
        '''compute the argmin of the trading strategy within a range of SMA's parameters: 
        (SMA_S:(start(10) , stop(50) , step(1)) , 
        SMA_L:(start(100), stop(252) , step(1))
         and compute the minimum value of  - objective function which is equivalent to the maximium of the objective function.        
        '''
        ranges = (SMAS_tuple , SMAL_tuple)
        obj_func  = lambda x : -1*self.update_and_run_strategy_performance_value(x)
        optimized_values = brute(obj_func , ranges)
        return optimized_values  ,   self.update_and_run_strategy_performance_value(optimized_values)
```

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
#plt.style.use("ggplot")
%config InlineBackend.figure_format = "retina"

from scipy.optimize import brute

import warnings
warnings.filterwarnings("ignore")
```

```python
# to Update our code to add instances to the SMA class:
class SMABacktester():
    ''' Class for the vectorized backtesting of SMA-based trading strategies:

    Attributes
    ==========
    symbol: str
        ticker symbol with which to work with
    SMA_S: int
        time window in days for shorter SMA
    SMA_L: int
        time window in days for longer SMA
    start: str
        start date for data retrieval
    end: str
        end date for data retrieval
    pct: float
        proportional trading cost
        
    Methods
    =======
    get_data:
        retrieves and prepares the data
        
    set_parameters:
        sets one or two new SMA parameters
        
    trading_strategy_performance_tabular:
        runs the backtest for the SMA-based strategy and return the KPI in a dataframe.

    trading_strategy_performance_values:
        runs the backtest for the SMA-based strategy and return the KPIs values.

    update_and_run_strategy_performance_table:
        Update the SMA parameter in the backtest trading strategy and display in a dataframe.

    update_and_run_strategy_performance_value:
        Update the SMA parameter in the backtest trading strategy and display the values.
         
    plot_results:
        plots the performance of the strategy compared to buy and hold.

    optimization_trading_parameters:
        compute the argmin of the trading strategy within a range of SMA's parameters. 
        '''
    def __init__(self , symbol:str , SMA_S:int , SMA_L:int , start:str , end:str , pct:float):
        self.symbol = symbol
        self.SMA_S = SMA_S
        self.SMA_L = SMA_L
        self.start = start
        self.end = end
        self.pct = pct
        self.results = None
        self.get_data()
        
    def __repr__(self):
        rep = "SMABacktester(symbol = {}, SMA_S = {}, SMA_L = {}, start = {}, end = {} , pct = {})"
        return rep.format(self.symbol, self.SMA_S, self.SMA_L, self.start, self.end , self.pct)     

    def get_data(self):
        '''Retreives and Prepares the data.
        '''
        raw = pd.read_csv("forex_pairs.csv" , parse_dates=["Date"] , index_col=["Date"])
        raw = raw[self.symbol].to_frame().dropna()
        raw = raw.loc[self.start : self.end].copy()
        raw.columns = ["price"]
        raw["SMA_S"] = raw["price"].rolling(window = self.SMA_S).mean()
        raw["SMA_L"] = raw["price"].rolling(window = self.SMA_L).mean()
        self.data = raw
        return raw

    def set_parameters(self , new_SMA_S = None, new_SMA_L = None):
        '''Updates SMA parameters and resp. time series.
        '''
        if new_SMA_S is not None:
            self.SMA_S = new_SMA_S
            self.data["SMA_S"] = self.data["price"].rolling(window = self.SMA_S).mean()

        if new_SMA_L is not None:
            self.SMA_L = new_SMA_L        
            self.data["SMA_L"] = self.data["price"].rolling(window = self.SMA_L).mean()

    def trading_strategy_performance_tabular(self):
        '''Display the performance of the trading strategy vs (Buy & Hold) in a dataframe.
        '''
        data = self.data.dropna().copy()
        data["position"] = np.where(data["SMA_S"] > data["SMA_L"] , +1 , -1)
        data["trades"] = data["position"].diff().fillna(0).abs() 
        data["log_ret"] = np.log(data["price"].div(data["price"].shift(1)))
        data["strategy"] = data["position"].shift(1).multiply(data["log_ret"]) - data["trades"]*self.pct
        data["cum_log_ret"] = np.exp(data["log_ret"].cumsum())
        data["cum_strategy"] = np.exp(data["strategy"].cumsum())

        perf = data["cum_strategy"].iloc[-1] # absolute performance
        cum_log_ret_perf = data["cum_log_ret"].iloc[-1] # Performance of the cummulative log returns
        out_perf = perf - cum_log_ret_perf  # out performance
        
        #pd.options.display.float_format = '{:.2f}%'.format
        self.KPI = pd.DataFrame({"Absolute Performance of Strategy": [perf] ,
                            "Cummulative Log Returns Performance" : [cum_log_ret_perf] ,
                            "Out Performance": [out_perf]} , 
                            index = ["results {}|{}".format(self.SMA_S , self.SMA_L)])
        
        self.KPI = self.KPI.T.style.format("{:.2%}")
        self.results = data        
        return self.KPI

    def trading_strategy_performance_values(self):
        '''Display the values of the cummulative performance of the trading strategy compared to Buy & Hold strategy.
        '''
        data = self.data.dropna().copy()
        data["position"] = np.where(data["SMA_S"] > data["SMA_L"] , +1 , -1)
        data["trades"] = data["position"].diff().fillna(0).abs()
        
        data["log_ret"] = np.log(data["price"].div(data["price"].shift(1)))
        data["strategy"] = data["position"].shift(1).multiply(data["log_ret"]) - data["trades"] * self.pct
        data["cum_log_ret"] = np.exp(data["log_ret"].cumsum())
        data["cum_strategy"] = np.exp(data["strategy"].cumsum())

        perf = data["cum_strategy"].iloc[-1] # absolute performance
        cum_log_ret_perf = data["cum_log_ret"].iloc[-1] # Performance of the cummulative log returns
        out_perf = perf - cum_log_ret_perf  # out performance
        
        #pd.options.display.float_format = '{:.2f}%'.format
        self.KPI = pd.DataFrame({"Absolute Performance of Strategy": [perf] ,
                            "Cummulative Log Returns Performance" : [cum_log_ret_perf] ,
                            "Out Performance": [out_perf]} , 
                            index = ["results {}|{}".format(self.SMA_S , self.SMA_L)])
        self.KPI = self.KPI.T
        self.results = data        
        return self.KPI.loc["Absolute Performance of Strategy",:].values

    def plot_results(self):
        '''Plot the cummulative returns of the trading strategy vs the cummulative log returns of the buy&hold strategy.
        '''
        if self.results is None:
            print("There is nothing to display! Run a Strategy :-)")
        else:
            self.results.iloc[:,-2:].plot()
            plt.title(f"Trading Strategy VS Buy&Hold Strategy \n Ticker: {self.symbol} | SMA_S: {self.SMA_S} | SMA_L: {self.SMA_L}")
            plt.xlabel("Date")
            plt.ylabel("Value")
            plt.tight_layout()
            plt.autoscale(enable = True, 
                          axis = "x", 
                          tight = True)
            plt.legend(labels = ["Buy&Hold Strategy" , "Trading Strategy"], 
                       bbox_to_anchor = (1,1) ,
                       facecolor = "peachpuff", 
                       frameon = True ,
                       fancybox = True , 
                       shadow = True)
            plt.show()

    def update_and_run_strategy_performance_table(self , SMA:tuple):
        '''Update the SMA parameter in the backtest trading strategy and return the dataframe of the trading strategy within the new parameters.
        '''
        self.set_parameters(int(SMA[0]) , int(SMA[1]))
        return self.trading_strategy_performance_tabular()

    def update_and_run_strategy_performance_value(self , SMA:tuple):
        '''Update the SMA parameter in the backtest trading strategy and return the value of the trading strategy within the new parameters.
           It will be used to find the final optimal value by using the brute function from scipy.optimize
        '''
        self.set_parameters(int(SMA[0]) , int(SMA[1]))
        return self.trading_strategy_performance_values()

    def optimization_trading_parameters(self , SMAS_tuple:tuple , SMAL_tuple:tuple):
        '''compute the argmin of the trading strategy within a range of SMA's parameters: 
        (SMA_S:(start(10) , stop(50) , step(1)) , 
        SMA_L:(start(100), stop(252) , step(1))
         and compute the minimum value of  - objective function which is equivalent to the maximium of the objective function.        
        '''
        ranges = (SMAS_tuple , SMAL_tuple)
        obj_func  = lambda x : -1*self.update_and_run_strategy_performance_value(x)
        optimized_values = brute(obj_func , ranges)
        return optimized_values  ,   self.update_and_run_strategy_performance_value(optimized_values)
```
